```python
%%sql
SELECT co.Continent, AVG(ci.population)
FROM country co
LEFT JOIN city ci ON (ci.CountryCode = co.Code)
WHERE (co.Population > 10000000)
GROUP BY co.Continent;
```

```python
%%sql

SELECT c.Region, l.Language
FROM countrylanguage l
LEFT JOIN country c ON (c.Code = l.CountryCode)
WHERE (c.Population > 50000000) AND (c.Continent = "Asia") AND (c.Region = "Middle East");
```

#### 24

```python
%%sql
USE IMDb;
```

```python
%%sql

SELECT AVG(r.average_rating), m.start_year
FROM Title_ratings r
INNER JOIN Titles m ON (m.title_id = r.title_id)
INNER JOIN Title_genres g ON (m.title_id = g.title_id)
WHERE (g.genre = "Action") AND (m.title_type = "movie") AND (m.runtime_minutes > 90)
GROUP BY m.start_year
ORDER BY m.start_year;
```

#### 25

```python
%%sql

SELECT n.name_, COUNT(m.primary_title) AS Moviecount
FROM (
SELECT * FROM Names_ LIMIT 1000
) AS n
INNER JOIN Had_role r ON (n.name_id = r.name_id)
LEFT JOIN Titles m ON (m.title_id = r.title_id)
WHERE (m.title_type = "movie")
GROUP BY n.name_
ORDER BY Moviecount DESC
LIMIT 5;
```

#### 26

```python
%%sql
SELECT  n.name_, AVG(r.average_rating)
FROM Directors d
INNER JOIN Names_ n ON (d.name_id = n.name_id)
INNER JOIN Title_genres g ON (d.title_id = g.title_id)
INNER JOIN Titles t ON (t.title_id = d.title_id)
INNER JOIN Title_ratings r ON (t.title_id = r.title_id)
WHERE (t.title_type = "movie") AND (g.genre = "Comedy") AND (t.start_year > 2000) AND (r.num_votes > 100000)
GROUP BY n.name_
HAVING (COUNT(t.title_id) > 5)
ORDER BY n.name_;
```

```python
%%sql

SELECT g.genre, AVG(t.runtime_minutes) AS avgtime, COUNT(t.title_id) AS movie_count
FROM Titles t
INNER JOIN Title_genres g ON (g.title_id = t.title_id)
WHERE (t.start_year > 2010) AND (t.title_type = "movie")
GROUP BY g.genre
HAVING (avgtime > 90) AND (movie_count > 10)
ORDER BY g.genre;
```

```python
%%sql
SELECT *
FROM (
    SELECT *
    FROM Names_
    WHERE (Names_.name_ = 'Federico Fellini')
) AS n
LEFT JOIN Directors d ON (d.name_id = n.name_id)
LEFT JOIN Had_role r ON (d.name_id = r.name_id)
ORDER BY d.title_id;
```

```python
%%sql
SELECT n.name_, d.title_id
FROM (
    SELECT *
    FROM Names_
    WHERE (Names_.name_ = 'Federico Fellini')
) AS n
LEFT JOIN Directors d ON (d.name_id = n.name_id)
UNION
SELECT n.name_, r.title_id
FROM (
    SELECT *
    FROM Names_
    WHERE (Names_.name_ = 'Federico Fellini')
) AS n
LEFT JOIN Had_role r ON (n.name_id = r.name_id);
```

```python
%%sql
SELECT n.name_, pm.title_id
FROM (
    SELECT *
    FROM Names_ n_inner
    WHERE n_inner.name_ = 'Federico Fellini'
) AS n
LEFT JOIN (
    SELECT d.name_id, d.title_id
    FROM Directors d
    UNION
    SELECT r.name_id, r.title_id
    FROM Had_role r
) AS pm ON (pm.name_id = n.name_id);
```
